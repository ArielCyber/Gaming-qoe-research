```python
import cv2
img = cv2.imread("/Users/<user>/PhD/qoe/video-ping/beni-frames/frame_718.jpg")
# create a yellow mask
lower_yellow = (0, 120, 120)
upper_yellow = (255, 255, 255)
mask = cv2.inRange(img, lower_yellow, upper_yellow)

# create a mask to extract only the yellow colored objects
yellow_mask = cv2.inRange(img, lower_yellow, upper_yellow)

# apply the mask to the image
yellow_result = cv2.bitwise_and(img, img, mask=yellow_mask)

# save the image to file
cv2.imwrite("yellow_result.jpg", yellow_result)
```
