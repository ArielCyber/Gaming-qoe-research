```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread('/Users/<user>/PhD/github/qoe-research/output/56/f_7141_raw.jpg')

# Convert the image from BGR to HSV color space
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Define the range of green color in HSV
# lower_green = np.array([20, 20, 120])  # Adjust these values based on your specific case
lower_none = np.array([0, 0, 0])
upper_none = np.array([255, 255, 255])

lower_yellow = np.array([0, 100, 100])  
upper_yellow = np.array([30, 255, 255])

lower_green = np.array([40, 40, 40])  
upper_green = np.array([80, 255, 255])

lower_red = np.array([40, 100, 100]) 
upper_red = np.array([255, 255, 255])

lower_blue = np.array([100, 50, 50])  
upper_blue = np.array([140, 255, 150])

for low_bound, upper_bound, title in [
    (lower_none, upper_none, 'No Filter'),
    (lower_blue, upper_blue, 'Blue Filter'),
    (lower_red, upper_red, 'Red Filter'),
    (lower_green, upper_green, 'Green Filter'),
    (lower_yellow, upper_yellow, 'Yellow Filter')]:
    # Create a mask using inRange function
    mask = cv2.inRange(hsv, low_bound, upper_bound)

    # Apply the mask to the original image
    result = cv2.bitwise_and(image, image, mask=mask)

    result2 = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
    # result2 = result[:,:,::-1]
    plt.title(title)
    plt.imshow(result2)
    plt.show()
```
